```python
# ============================================================
# @title ⚙️ Preprocess Dataset    22
# ============================================================

import os
import subprocess
import time

print("=" * 60)

# ------------------------------------------------------------
# Project
# ------------------------------------------------------------

PROJECT_PATH = "/content/Applio"

if not os.path.isdir(PROJECT_PATH):
    raise RuntimeError(
        f"Applio project was not found at:\n{PROJECT_PATH}"
    )

os.chdir(PROJECT_PATH)

# ------------------------------------------------------------
# v5 configuration
# ------------------------------------------------------------

# SAMPLE_RATE is defined in Cell 3 of v5.
# After a runtime restart, use the same value explicitly here.
SAMPLE_RATE = "40000"

print(f"Working directory : {os.getcwd()}")
print(f"Model             : {MODEL_NAME}")
print(f"Dataset           : {DATASET_PATH}")
print(f"Sample rate       : {SAMPLE_RATE}")

print("=" * 60)

# ------------------------------------------------------------
# Check core.py
# ------------------------------------------------------------

CORE_PATH = os.path.join(PROJECT_PATH, "core.py")

if not os.path.isfile(CORE_PATH):
    raise RuntimeError(
        f"core.py was not found at:\n{CORE_PATH}"
    )

# ------------------------------------------------------------
# Check dataset
# ------------------------------------------------------------

if not os.path.isdir(DATASET_PATH):
    raise RuntimeError(
        f"Dataset folder was not found at:\n{DATASET_PATH}"
    )

# ------------------------------------------------------------
# Experiment path
# ------------------------------------------------------------

EXP_PATH = os.path.join(
    PROJECT_PATH,
    "logs",
    MODEL_NAME
)

SLICED_PATH = os.path.join(
    EXP_PATH,
    "sliced_audios_16k"
)

# ------------------------------------------------------------
# Check existing preprocessing
# ------------------------------------------------------------

existing_files = []

if os.path.isdir(SLICED_PATH):
    existing_files = [
        f
        for f in os.listdir(SLICED_PATH)
        if f.lower().endswith(".wav")
    ]

# ------------------------------------------------------------
# Run preprocessing only if needed
# ------------------------------------------------------------

if existing_files:

    print("Existing preprocessed audio detected.")
    print(
        f"sliced_audios_16k files : {len(existing_files)}"
    )
    print("Preprocess skipped.")

else:

    print("New project detected.")
    print("Running preprocessing...")

    start_time = time.time()

    command = [
        "python",
        CORE_PATH,
        "preprocess",
        "--model_name",
        MODEL_NAME,
        "--dataset_path",
        DATASET_PATH,
        "--sample_rate",
        str(SAMPLE_RATE),
        "--cut_preprocess",
        "Automatic",
    ]

    print("=" * 60)
    print("Running preprocessing...")
    print("=" * 60)

    result = subprocess.run(
        command,
        cwd=PROJECT_PATH,
        text=True,
    )

    elapsed = time.time() - start_time

    if result.returncode != 0:
        raise RuntimeError(
            f"Preprocessing failed with exit code "
            f"{result.returncode}."
        )

    print("=" * 60)
    print(
        f"Preprocess completed in {elapsed:.2f} seconds."
    )
    print("=" * 60)

# ------------------------------------------------------------
# Final verification
# ------------------------------------------------------------

if not os.path.isdir(SLICED_PATH):
    raise RuntimeError(
        "Preprocessing output folder was not created:\n"
        f"{SLICED_PATH}"
    )

final_files = [
    f
    for f in os.listdir(SLICED_PATH)
    if f.lower().endswith(".wav")
]

print("=" * 60)
print("Preprocessing verification")
print("=" * 60)
print(f"sliced_audios_16k : {len(final_files)}")

if len(final_files) == 0:
    raise RuntimeError(
        "Preprocessing completed but no WAV files were found "
        "inside sliced_audios_16k."
    )

print("=" * 60)
print("✓ Preprocessing stage completed successfully")
print("=" * 60)
```

```python
# ============================================================
# @title 🎼 Extract F0     23
# ============================================================

import os
import subprocess
import sys
import json

print("=" * 60)
print("Extracting F0...")
print("=" * 60)

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

PROJECT_PATH = "/content/Applio"

MODEL_NAME = "xyz"
SAMPLE_RATE = "40000"
F0_METHOD = "rmvpe"

CPU_CORES = 2
INCLUDE_MUTES = 2
EMBEDDER_MODEL = "contentvec"

CORE_PATH = os.path.join(
    PROJECT_PATH,
    "core.py"
)

EXP_PATH = os.path.join(
    PROJECT_PATH,
    "logs",
    MODEL_NAME
)

F0_PATH = os.path.join(
    EXP_PATH,
    "f0"
)

F0_VOICED_PATH = os.path.join(
    EXP_PATH,
    "f0_voiced"
)

EXTRACTED_PATH = os.path.join(
    EXP_PATH,
    "extracted"
)

SLICED_16K_PATH = os.path.join(
    EXP_PATH,
    "sliced_audios_16k"
)

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print(f"Project      : {PROJECT_PATH}")
print(f"Model        : {MODEL_NAME}")
print(f"F0 Method    : {F0_METHOD}")
print(f"Sample Rate  : {SAMPLE_RATE}")
print(f"CPU Cores    : {CPU_CORES}")
print(f"Embedder     : {EMBEDDER_MODEL}")

print("=" * 60)

# ------------------------------------------------------------
# Safety checks
# ------------------------------------------------------------

if not os.path.isfile(CORE_PATH):
    raise RuntimeError(
        f"core.py not found:\n{CORE_PATH}"
    )

if not os.path.isdir(EXP_PATH):
    raise RuntimeError(
        f"Experiment folder not found:\n{EXP_PATH}"
    )

if not os.path.isdir(SLICED_16K_PATH):
    raise RuntimeError(
        f"sliced_audios_16k not found:\n{SLICED_16K_PATH}\n"
        "Run Cell 22 first."
    )

wav_files = sorted(
    f for f in os.listdir(SLICED_16K_PATH)
    if f.lower().endswith(".wav")
)

print(
    f"sliced_audios_16k files : {len(wav_files)}"
)

if len(wav_files) == 0:
    raise RuntimeError(
        "No WAV files found in sliced_audios_16k."
    )

# ------------------------------------------------------------
# Prepare output folders
# ------------------------------------------------------------

os.makedirs(F0_PATH, exist_ok=True)
os.makedirs(F0_VOICED_PATH, exist_ok=True)
os.makedirs(EXTRACTED_PATH, exist_ok=True)

# ------------------------------------------------------------
# Build command
# ------------------------------------------------------------

command = [
    sys.executable,
    CORE_PATH,
    "extract",

    "--model_name",
    MODEL_NAME,

    "--f0_method",
    F0_METHOD,

    "--cpu_cores",
    str(CPU_CORES),

    "--gpu",
    "-",

    "--sample_rate",
    SAMPLE_RATE,

    "--embedder_model",
    EMBEDDER_MODEL,

    "--include_mutes",
    str(INCLUDE_MUTES),
]

print("=" * 60)
print("Running Applio F0 / embedding extraction")
print("=" * 60)

print(
    "Command:",
    " ".join(command)
)

print("=" * 60)

# ------------------------------------------------------------
# Run extraction
# ------------------------------------------------------------

result = subprocess.run(
    command,
    cwd=PROJECT_PATH,
    text=True
)

# ------------------------------------------------------------
# Check process result
# ------------------------------------------------------------

print("=" * 60)

if result.returncode != 0:
    raise RuntimeError(
        "Applio extraction failed.\n"
        f"Exit code: {result.returncode}"
    )

print("Applio extraction process completed.")

# ------------------------------------------------------------
# Verify F0 output
# ------------------------------------------------------------

f0_files = [
    f
    for f in os.listdir(F0_PATH)
    if f.endswith(".npy")
]

f0_voiced_files = [
    f
    for f in os.listdir(F0_VOICED_PATH)
    if f.endswith(".npy")
]

feature_files = [
    f
    for f in os.listdir(EXTRACTED_PATH)
    if f.endswith(".npy")
]

print("=" * 60)
print("Checking extraction results")
print("=" * 60)

print(
    f"F0 files        : {len(f0_files)}"
)

print(
    f"F0 voiced files : {len(f0_voiced_files)}"
)

print(
    f"Feature files   : {len(feature_files)}"
)

print("=" * 60)

# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

if len(f0_files) == 0:
    raise RuntimeError(
        "Extraction process returned successfully, "
        "but no F0 files were created."
    )

if len(f0_voiced_files) == 0:
    raise RuntimeError(
        "Extraction process returned successfully, "
        "but no f0_voiced files were created."
    )

print(
    "✓ F0 extraction completed successfully."
)

print("=" * 60)
```

```python

# ============================================================
#@title 🎧 Extract Features     24
# ============================================================

import os
import json
from random import shuffle

print("="*60)
print("Checking training files...")
print("="*60)


MODEL_PATH = f"/content/Applio/logs/{MODEL_NAME}"


sliced_dir = os.path.join(
    MODEL_PATH,
    "sliced_audios"
)

feature_dir = os.path.join(
    MODEL_PATH,
    "extracted"
)

f0_dir = os.path.join(
    MODEL_PATH,
    "f0"
)

f0nsf_dir = os.path.join(
    MODEL_PATH,
    "f0_voiced"
)


def get_names(path):

    if not os.path.exists(path):
        return set()

    return set(
        x.split(".")[0]
        for x in os.listdir(path)
    )


wav_names = get_names(sliced_dir)
feature_names = get_names(feature_dir)
f0_names = get_names(f0_dir)
f0nsf_names = get_names(f0nsf_dir)


print("sliced_audios :", len(wav_names))
print("features      :", len(feature_names))
print("f0            :", len(f0_names))
print("f0_voiced     :", len(f0nsf_names))


names = (
    wav_names
    &
    feature_names
    &
    f0_names
    &
    f0nsf_names
)


print("="*60)
print("Matched training samples :", len(names))
print("="*60)


if len(names) == 0:

    raise RuntimeError(
        """
No matched training samples found.

Check:
sliced_audios
extracted
f0
f0_voiced

Training cannot continue.
"""
    )


filelist = []

for name in names:

    sid = name.split("_")[0]

    wav = os.path.relpath(
        os.path.join(sliced_dir, name + ".wav")
    )

    feat = os.path.relpath(
        os.path.join(feature_dir, name + ".npy")
    )

    f0 = os.path.relpath(
        os.path.join(f0_dir, name + ".wav.npy")
    )

    f0nsf = os.path.relpath(
        os.path.join(f0nsf_dir, name + ".wav.npy")
    )


    filelist.append(
        f"{wav}|{feat}|{f0}|{f0nsf}|{sid}"
        .replace("\\","/")
    )


shuffle(filelist)


filelist_path = os.path.join(
    MODEL_PATH,
    "filelist.txt"
)


with open(filelist_path,"w") as f:
    f.write("\n".join(filelist))


print("="*60)
print("filelist created")
print("Samples :",len(filelist))
print(filelist[:3])
print("="*60)
```

```python
# ============================================================
# @title 📦 Package Final Model
# ============================================================
# این سلول فقط فایل‌های نهایی مدل آموزش‌دیده را بررسی و
# در یک فایل ZIP قابل انتقال بسته‌بندی می‌کند.
# هیچ مرحله‌ای از Preprocess / F0 / Feature Extraction / Training
# دوباره اجرا نمی‌شود.
# ============================================================

import os
import zipfile
from datetime import datetime

# ============================================================
# Configuration
# ============================================================

PROJECT_PATH = "/content/Applio"
MODEL_NAME = "xyz"
EXP_PATH = os.path.join(PROJECT_PATH, "logs", MODEL_NAME)

FINAL_MODEL = os.path.join(
    EXP_PATH,
    f"{MODEL_NAME}_5e_195s.pth"
)

INDEX_FILE = os.path.join(
    EXP_PATH,
    f"{MODEL_NAME}.index"
)

CONFIG_FILE = os.path.join(
    EXP_PATH,
    "config.json"
)

MODEL_INFO_FILE = os.path.join(
    EXP_PATH,
    "model_info.json"
)

FILELIST_FILE = os.path.join(
    EXP_PATH,
    "filelist.txt"
)

ZIP_PATH = os.path.join(
    EXP_PATH,
    f"{MODEL_NAME}_final_package.zip"
)

# ============================================================
# Header
# ============================================================

print("=" * 60)
print("Packaging final trained model")
print("=" * 60)

print(f"Project : {PROJECT_PATH}")
print(f"Model   : {MODEL_NAME}")
print(f"Output  : {EXP_PATH}")
print("=" * 60)

# ============================================================
# Required files
# ============================================================

required_files = {
    "Final Model": FINAL_MODEL,
    "Index": INDEX_FILE,
    "Config": CONFIG_FILE,
    "Model Info": MODEL_INFO_FILE,
    "Filelist": FILELIST_FILE,
}

print("Checking final files...")
print("=" * 60)

missing_files = []

for label, path in required_files.items():
    if os.path.isfile(path):
        size_mb = os.path.getsize(path) / (1024 * 1024)
        print(f"✓ {label:<12}: {os.path.basename(path)} "
              f"({size_mb:.2f} MB)")
    else:
        print(f"✗ {label:<12}: NOT FOUND")
        missing_files.append(label)

print("=" * 60)

# ============================================================
# Stop if required files are missing
# ============================================================

if missing_files:
    raise RuntimeError(
        "Final package cannot be created.\n"
        "Missing required files: "
        + ", ".join(missing_files)
    )

# ============================================================
# Remove previous ZIP if it exists
# ============================================================

if os.path.exists(ZIP_PATH):
    os.remove(ZIP_PATH)

# ============================================================
# Create ZIP package
# ============================================================

print("Creating final ZIP package...")
print("=" * 60)

with zipfile.ZipFile(
    ZIP_PATH,
    mode="w",
    compression=zipfile.ZIP_DEFLATED
) as zipf:

    for label, path in required_files.items():
        zipf.write(
            path,
            arcname=os.path.basename(path)
        )
        print(f"✓ Added: {os.path.basename(path)}")

# ============================================================
# Verification
# ============================================================

print("=" * 60)
print("Verifying ZIP package...")
print("=" * 60)

with zipfile.ZipFile(ZIP_PATH, mode="r") as zipf:
    zip_contents = zipf.namelist()

for label, path in required_files.items():
    filename = os.path.basename(path)

    if filename in zip_contents:
        print(f"✓ Verified: {filename}")
    else:
        raise RuntimeError(
            f"ZIP verification failed: {filename} not found."
        )

# ============================================================
# Final information
# ============================================================

zip_size_mb = os.path.getsize(ZIP_PATH) / (1024 * 1024)

print("=" * 60)
print("FINAL PACKAGE CREATED SUCCESSFULLY")
print("=" * 60)

print(f"ZIP File : {ZIP_PATH}")
print(f"Size     : {zip_size_mb:.2f} MB")
print(f"Created  : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("=" * 60)

print("Contents:")
for filename in zip_contents:
    print(f"  - {filename}")

print("=" * 60)
print("✓ Model package is ready for transfer / backup.")
print("=" * 60)
```
