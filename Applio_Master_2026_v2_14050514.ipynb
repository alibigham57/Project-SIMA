```python
# ============================================================
#@title ♻ Restore Complete Project   10
# ============================================================

import os
import shutil

os.makedirs(PROJECT_PATH, exist_ok=True)

for item in os.listdir(BACKUP_PATH):

    src = os.path.join(BACKUP_PATH, item)
    dst = os.path.join(PROJECT_PATH, item)

    if os.path.isdir(src):

        if os.path.exists(dst):
            shutil.rmtree(dst)

        shutil.copytree(src, dst)

        print("Folder :", item)

    else:

        shutil.copy2(src, dst)

        print("File   :", item)

print("="*60)
print("Project Restored Successfully")
print("="*60)
```

```python
# ============================================================
#@title 🚀 Train Model    17
# ============================================================

import glob
import os

RUN_TRAIN = True #@param {type:"boolean"}

resume = len(glob.glob(PROJECT_PATH+"/G_*.pth")) > 0

print("Resume :", resume)

command = f"""
python core.py train \
--model_name "{MODEL_NAME}" \
--sample_rate {SAMPLE_RATE} \
--total_epoch {TOTAL_EPOCH} \
--batch_size {BATCH_SIZE} \
--save_every_epoch {SAVE_EVERY_EPOCH}
"""

print(command)

if not resume:
    raise RuntimeError(
        "Checkpoint not found. Training would start from scratch."
    )

if RUN_TRAIN:
    !$command
```

```python
# ============================================================
#@title 💾 Backup Complete Project
# ============================================================

import os
import shutil

print("="*70)

print("Synchronizing Project...")

print("="*70)

os.makedirs(BACKUP_PATH, exist_ok=True)

copied = 0

for item in os.listdir(PROJECT_PATH):

    src = os.path.join(PROJECT_PATH, item)

    dst = os.path.join(BACKUP_PATH, item)

    if os.path.isdir(src):

        if os.path.exists(dst):

            shutil.rmtree(dst)

        shutil.copytree(src, dst)

        print("Folder :", item)

    else:

        shutil.copy2(src, dst)

        print("File   :", item)

    copied += 1

print("="*70)

print("Backup Completed")

print("Items :", copied)

print("="*70)
```
